```python
user = user.dropna().reset_index(drop=True) #drops any row that has a NaN, these rows are broken, not just incomplete, and they're under 1% of the data. Fixing them by hand isn't worth it.
print("Real NaN values left:", user.isna().sum().sum())
print("Rows:", len(user))

#Order matters: this runs before the next section turns empty reviews into NaN. 
#If it ran after, those empty reviews would be deleted too, and we want to keep them.
```

# Find Hidden Missing Values e.g "None"
### Pandas can't tell that the text "None" means missing, so it counts "None" as a real reviewer name or date

```python
text_cols = user.select_dtypes(include=['object', 'str']).columns
placeholders = ['', 'nan', 'none', 'null', 'n/a', 'na', '-', '?']

for col in text_cols:
    s = user[col].astype(str).str.strip().str.strip("'\"").str.strip().str.lower()
    found = s[s.isin(placeholders)].value_counts()
    if len(found):
        print(col, found.to_dict())
```

```python
s = user['reviewer'].str.strip().str.strip("'\"").str.strip().str.lower()
user.loc[s.isin(['nan', 'na', 'none']), 'reviewer'] = np.nan

s = user['Rev'].str.strip().str.strip("'\"").str.strip()
user.loc[s.isin(['', '?']), 'Rev'] = np.nan
```

### Thumbs

```python
for col in ['thumbsUp', 'thumbsTot']:
    num = pd.to_numeric(user[col], errors='coerce')
    print(col, "text values:", (num.isna() & user[col].notna()).sum())
```

### info() showed the thumbs columns as object, which means at least one value isn't a number, and dropna() only removes empty cells, not text, so we count the text values before converting to see exactly what gets dropped.

```python
for col in ['thumbsUp', 'thumbsTot']:
    user[col] = pd.to_numeric(user[col], errors='coerce')

user = user.dropna(subset=['thumbsUp', 'thumbsTot']).reset_index(drop=True)
user[['thumbsUp', 'thumbsTot']] = user[['thumbsUp', 'thumbsTot']].astype(int)

print("Rows:", len(user))
print("thumbsUp > thumbsTot:", (user['thumbsUp'] > user['thumbsTot']).sum())
user[['idvscore', 'thumbsUp', 'thumbsTot']].dtypes
```

### thumbsTot had 1 value containing scraped text; that row was dropped (315,957 → 315,956). Both columns are now integers and thumbsUp never exceeds thumbsTot.

### Dates

```python
user['dateP'].sample(10, random_state=1)
#This cell shows 10 random dates so you can see what format they're in before converting them.
```

### Same quote marks as in the expert file, so we strip them before converting.

```python
user['dateP'] = pd.to_datetime(
    user['dateP'].str.strip().str.strip("'"),
    format='%b %d, %Y'
)
```

```python
print(user['dateP'].isna().sum()) #number of missing dates
print(user['dateP'].min(), user['dateP'].max())
user['dateP'].dtype
```

### Reviewer and Rev still have the quote marks around them.
### Some reviews were cut off by the scraper, so they have an opening quote but no closing one. That's why we remove the first and last quote separately instead of only matching pairs like in the expert file.

```python
for col in ['reviewer', 'Rev']:
    user[col] = (
        user[col]
        .str.strip()
        .str.strip("'\"")
        .str.replace(r"\\[rnt]", " ", regex=True)
        .str.replace(r"\s{2,}", " ", regex=True)
        .str.strip()
    )
```

```python
for col in ['reviewer', 'Rev']:
    print(col, user[col].str.match(r"^['\"]").sum())
```

### Add id_user_review as the primary key, after all rows have been removed so the IDs are 1..n with no gaps.

```python
if 'id_user_review' not in user.columns:  # so re-running doesn't crash
    user.insert(0, 'id_user_review', range(1, len(user) + 1))

user[['id_user_review', 'url', 'reviewer']].head()
```
